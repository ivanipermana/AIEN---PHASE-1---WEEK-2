# Demo singkat Pandas
Empat kasus sintetis terpisah dari dataset challenge. Instruktur menjalankan demo sebelum challenge.

```python
import pandas as pd
toy = pd.DataFrame({'version':['X','X','Y','Y'], 'passed':[1,0,1,1], 'seconds':[1.,3.,2.,4.]})
display(toy)
```

```python
toy.groupby('version').agg(n=('passed','count'), pass_rate=('passed','mean'), median_s=('seconds','median'))
```

Mean label 0/1 menghasilkan proporsi. Kalikan 100 jika tampilan dalam persen. Median dari dua nilai adalah rata-rata kedua nilai tengah.
